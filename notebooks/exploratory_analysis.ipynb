# FanSphere AI — Exploratory Analysis

Quick scratchpad for inspecting pipeline outputs before they are formalised into the Power BI dashboard.

```python
import pandas as pd
from sqlalchemy import text
from src.config import get_engine

engine = get_engine()
```

## Top matches by engagement

```python
with engine.connect() as conn:
    df = pd.read_sql(text("SELECT * FROM v_executive_overview ORDER BY engagement_score DESC NULLS LAST LIMIT 10"), conn)
df
```

## Sentiment distribution

```python
with engine.connect() as conn:
    sent = pd.read_sql(text("SELECT sentiment_label, emotion, sentiment_score FROM fan_sentiment"), conn)
sent['sentiment_label'].value_counts()
```
